```python
import torch
from unsloth import FastLanguageModel
from transformers import pipeline, AutoTokenizer

model_name = "Psylo1226/llama-3.2-course-recommender-verB"

# ✅ Load model and tokenizer using FastLanguageModel
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name,
    max_seq_length=4096,     # Define max sequence length
    dtype=torch.float16,     # Efficient float16 dtype for GPU
    load_in_4bit=True        # Load with 4-bit precision for lower VRAM usage
)

# ✅ Prepare the model for inference (Unsloth specific)
model = FastLanguageModel.for_inference(model)

# ✅ Create the text-generation pipeline
pipe = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    torch_dtype=torch.float16,
    do_sample=True,
    temperature=0.7,
    top_p=0.9,
    max_new_tokens=1000,  # Ensure long enough responses
    pad_token_id=tokenizer.eos_token_id,  # Avoid padding issues
    eos_token_id=tokenizer.eos_token_id,  # Ensure natural stopping
)

print("✅ Model loaded successfully for inference on GPU!")
```

```python
# ✅ Define the course recommendation function
def recommend_course(topic):
    prompt = f"""### Instrukcja:
Zarekomenduj kurs w tematyce {topic}.

### Odpowiedź:
"""
    output = pipe(
        prompt,
        top_k=100,              # Limit to top 50 likely words
    )
    return output[0]['generated_text']


# ✅ Test the model
print(recommend_course("ISQTB Tester - poziom podstawowy"))
```
